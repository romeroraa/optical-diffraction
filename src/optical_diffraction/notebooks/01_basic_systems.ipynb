```python
# Compare different NA values for 750 nm
NAs = [0.5, 0.8, 1.2, 1.4]
systems = [OpticalSystem(750e-9, na, 1.52) for na in NAs]

for sys in systems:
    print(f"NA={sys.NA:.1f}: Diffraction limit = {sys.diffraction_limit*1e9:.0f} nm, "
          f"Rayleigh length = {sys.rayleigh_length*1e6:.1f} μm")
```
