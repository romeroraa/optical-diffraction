```python
import numpy as np
import matplotlib.pyplot as plt
import optical_diffraction as od

plt.rcParams['figure.figsize'] = (14, 6)
plt.rcParams['font.size'] = 11
plt.rcParams['lines.linewidth'] = 2.5

colors = {'scalar': '#1f77b4', 'vector': '#ff7f0e'}
```

```python
wavelength = 750e-9
n_medium = 1.3
NA_low = 0.3
NA_high = 1.1

system_low = od.OpticalSystem(
    wavelength=wavelength,
    numerical_aperture=NA_low,
    refractive_index=n_medium
)

system_high = od.OpticalSystem(
    wavelength=wavelength,
    numerical_aperture=NA_high,
    refractive_index=n_medium
)

print(f"Low NA: z_R = {system_low.rayleigh_length*1e6:.2f} μm, r_Airy = {system_low.airy_radius*1e9:.0f} nm")
print(f"High NA: z_R = {system_high.rayleigh_length*1e6:.2f} μm, r_Airy = {system_high.airy_radius*1e9:.0f} nm")
```

```python
# Plot 1: On-Axis Profiles - NA = 0.3
print("Computing on-axis profiles for NA = 0.3...")

z_range = 4 * system_low.rayleigh_length
r_range = 4 * system_low.airy_radius

z_coords_s, Ex_s, Ey_s, Ez_s = od.create_axial_profile(
    system_low, z_range=z_range, nz=401, r_constant=0.0, method='scalar'
)
z_coords_v, Ex_v, Ey_v, Ez_v = od.create_axial_profile(
    system_low, z_range=z_range, nz=401, r_constant=0.0, method='vector'
)

r_coords_s, Ex_s2, Ey_s2, Ez_s2 = od.create_transverse_profile(
    system_low, r_range=r_range, nr=401, z_constant=0.0, method='scalar'
)
r_coords_v, Ex_v2, Ey_v2, Ez_v2 = od.create_transverse_profile(
    system_low, r_range=r_range, nr=401, z_constant=0.0, method='vector'
)

I_axial_scalar = np.abs(Ex_s)**2 + np.abs(Ey_s)**2 + np.abs(Ez_s)**2
I_axial_vector = np.abs(Ex_v)**2 + np.abs(Ey_v)**2 + np.abs(Ez_v)**2
I_trans_scalar = np.abs(Ex_s2)**2 + np.abs(Ey_s2)**2 + np.abs(Ez_s2)**2
I_trans_vector = np.abs(Ex_v2)**2 + np.abs(Ey_v2)**2 + np.abs(Ez_v2)**2

I_axial_scalar = I_axial_scalar / np.max(I_axial_scalar)
I_axial_vector = I_axial_vector / np.max(I_axial_vector)
I_trans_scalar = I_trans_scalar / np.max(I_trans_scalar)
I_trans_vector = I_trans_vector / np.max(I_trans_vector)

z_um = z_coords_s * 1e6
r_um = r_coords_s * 1e6

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.plot(z_um, I_axial_scalar, color=colors['scalar'], label='Scalar', linewidth=2.5)
ax1.plot(z_um, I_axial_vector, color=colors['vector'], label='Vector (Richards-Wolf)', linewidth=2.5)
ax1.set_xlabel('z (μm)')
ax1.set_ylabel('Normalized Intensity')
ax1.set_title('Axial Profile (r = 0)')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(r_um, I_trans_scalar, color=colors['scalar'], label='Scalar', linewidth=2.5)
ax2.plot(r_um, I_trans_vector, color=colors['vector'], label='Vector (Richards-Wolf)', linewidth=2.5)
ax2.set_xlabel('r (μm)')
ax2.set_ylabel('Normalized Intensity')
ax2.set_title('Transverse Profile (z = 0)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.suptitle(f'On-Axis Profiles - NA = {NA_low} (750 nm, n = {n_medium})', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("Low NA results: Good agreement between scalar and vector (paraxial regime valid)")
```

```python
# Plot 2: On-Axis Profiles - NA = 1.1
print("Computing on-axis profiles for NA = 1.1...")

z_range = 4 * system_high.rayleigh_length
r_range = 4 * system_high.airy_radius

z_coords_s, Ex_s, Ey_s, Ez_s = od.create_axial_profile(
    system_high, z_range=z_range, nz=401, r_constant=0.0, method='scalar'
)
z_coords_v, Ex_v, Ey_v, Ez_v = od.create_axial_profile(
    system_high, z_range=z_range, nz=401, r_constant=0.0, method='vector'
)

r_coords_s, Ex_s2, Ey_s2, Ez_s2 = od.create_transverse_profile(
    system_high, r_range=r_range, nr=401, z_constant=0.0, method='scalar'
)
r_coords_v, Ex_v2, Ey_v2, Ez_v2 = od.create_transverse_profile(
    system_high, r_range=r_range, nr=401, z_constant=0.0, method='vector'
)

I_axial_scalar = np.abs(Ex_s)**2 + np.abs(Ey_s)**2 + np.abs(Ez_s)**2
I_axial_vector = np.abs(Ex_v)**2 + np.abs(Ey_v)**2 + np.abs(Ez_v)**2
I_trans_scalar = np.abs(Ex_s2)**2 + np.abs(Ey_s2)**2 + np.abs(Ez_s2)**2
I_trans_vector = np.abs(Ex_v2)**2 + np.abs(Ey_v2)**2 + np.abs(Ez_v2)**2

I_Ez_axial = np.abs(Ez_v)**2

I_axial_scalar = I_axial_scalar / np.max(I_axial_scalar)
I_axial_vector = I_axial_vector / np.max(I_axial_vector)
I_trans_scalar = I_trans_scalar / np.max(I_trans_scalar)
I_trans_vector = I_trans_vector / np.max(I_trans_vector)

z_um = z_coords_s * 1e6
r_um = r_coords_s * 1e6

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.plot(z_um, I_axial_scalar, color=colors['scalar'], label='Scalar', linewidth=2.5)
ax1.plot(z_um, I_axial_vector, color=colors['vector'], label='Vector (Richards-Wolf)', linewidth=2.5)
ax1.set_xlabel('z (μm)')
ax1.set_ylabel('Normalized Intensity')
ax1.set_title('Axial Profile (r = 0)')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(r_um, I_trans_scalar, color=colors['scalar'], label='Scalar', linewidth=2.5)
ax2.plot(r_um, I_trans_vector, color=colors['vector'], label='Vector (Richards-Wolf)', linewidth=2.5)
ax2.set_xlabel('r (μm)')
ax2.set_ylabel('Normalized Intensity')
ax2.set_title('Transverse Profile (z = 0)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.suptitle(f'On-Axis Profiles - NA = {NA_high} (750 nm, n = {n_medium})', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

focus_idx = len(z_coords_v) // 2
Ez_contribution = I_Ez_axial[focus_idx]
total_contribution = I_axial_vector[focus_idx]
Ez_percentage = (Ez_contribution / np.max(I_axial_vector)) * 100

print(f"High NA results: Significant differences - vector theory predicts sharper focusing")
print(f"Ez field contribution at focus: {Ez_percentage:.1f}% of total intensity")
```

```python
# Plot 3: Off-Center Profiles - NA = 0.3
print("Computing off-center profiles for NA = 0.3...")

r_offset = 1.5 * system_low.airy_radius
z_offset = 1.5 * system_low.rayleigh_length
z_range = 4 * system_low.rayleigh_length
r_range = 4 * system_low.airy_radius

z_coords_s, Ex_s, Ey_s, Ez_s = od.create_axial_profile(
    system_low, z_range=z_range, nz=401, r_constant=r_offset, method='scalar'
)
z_coords_v, Ex_v, Ey_v, Ez_v = od.create_axial_profile(
    system_low, z_range=z_range, nz=401, r_constant=r_offset, method='vector'
)

r_coords_s, Ex_s2, Ey_s2, Ez_s2 = od.create_transverse_profile(
    system_low, r_range=r_range, nr=401, z_constant=z_offset, method='scalar'
)
r_coords_v, Ex_v2, Ey_v2, Ez_v2 = od.create_transverse_profile(
    system_low, r_range=r_range, nr=401, z_constant=z_offset, method='vector'
)

I_axial_scalar = np.abs(Ex_s)**2 + np.abs(Ey_s)**2 + np.abs(Ez_s)**2
I_axial_vector = np.abs(Ex_v)**2 + np.abs(Ey_v)**2 + np.abs(Ez_v)**2
I_trans_scalar = np.abs(Ex_s2)**2 + np.abs(Ey_s2)**2 + np.abs(Ez_s2)**2
I_trans_vector = np.abs(Ex_v2)**2 + np.abs(Ey_v2)**2 + np.abs(Ez_v2)**2

I_axial_scalar = I_axial_scalar / np.max(I_axial_scalar)
I_axial_vector = I_axial_vector / np.max(I_axial_vector)
I_trans_scalar = I_trans_scalar / np.max(I_trans_scalar)
I_trans_vector = I_trans_vector / np.max(I_trans_vector)

z_um = z_coords_s * 1e6
r_um = r_coords_s * 1e6
r_offset_um = r_offset * 1e6
z_offset_um = z_offset * 1e6

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.plot(z_um, I_axial_scalar, color=colors['scalar'], label='Scalar', linewidth=2.5)
ax1.plot(z_um, I_axial_vector, color=colors['vector'], label='Vector (Richards-Wolf)', linewidth=2.5)
ax1.set_xlabel('z (μm)')
ax1.set_ylabel('Normalized Intensity')
ax1.set_title(f'Axial Profile (r = {r_offset_um:.2f} μm)')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(r_um, I_trans_scalar, color=colors['scalar'], label='Scalar', linewidth=2.5)
ax2.plot(r_um, I_trans_vector, color=colors['vector'], label='Vector (Richards-Wolf)', linewidth=2.5)
ax2.set_xlabel('r (μm)')
ax2.set_ylabel('Normalized Intensity')
ax2.set_title(f'Transverse Profile (z = {z_offset_um:.2f} μm)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.suptitle(f'Off-Center Profiles - NA = {NA_low} (750 nm, n = {n_medium})', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("Low NA off-center: Still good agreement, minor differences away from axis")
```

```python
# Plot 4: Off-Center Profiles - NA = 1.1
print("Computing off-center profiles for NA = 1.1...")

r_offset = 1.5 * system_high.airy_radius
z_offset = 1.5 * system_high.rayleigh_length
z_range = 4 * system_high.rayleigh_length
r_range = 4 * system_high.airy_radius

z_coords_s, Ex_s, Ey_s, Ez_s = od.create_axial_profile(
    system_high, z_range=z_range, nz=401, r_constant=r_offset, method='scalar'
)
z_coords_v, Ex_v, Ey_v, Ez_v = od.create_axial_profile(
    system_high, z_range=z_range, nz=401, r_constant=r_offset, method='vector'
)

r_coords_s, Ex_s2, Ey_s2, Ez_s2 = od.create_transverse_profile(
    system_high, r_range=r_range, nr=401, z_constant=z_offset, method='scalar'
)
r_coords_v, Ex_v2, Ey_v2, Ez_v2 = od.create_transverse_profile(
    system_high, r_range=r_range, nr=401, z_constant=z_offset, method='vector'
)

I_axial_scalar = np.abs(Ex_s)**2 + np.abs(Ey_s)**2 + np.abs(Ez_s)**2
I_axial_vector = np.abs(Ex_v)**2 + np.abs(Ey_v)**2 + np.abs(Ez_v)**2
I_trans_scalar = np.abs(Ex_s2)**2 + np.abs(Ey_s2)**2 + np.abs(Ez_s2)**2
I_trans_vector = np.abs(Ex_v2)**2 + np.abs(Ey_v2)**2 + np.abs(Ez_v2)**2

I_axial_scalar = I_axial_scalar / np.max(I_axial_scalar)
I_axial_vector = I_axial_vector / np.max(I_axial_vector)
I_trans_scalar = I_trans_scalar / np.max(I_trans_scalar)
I_trans_vector = I_trans_vector / np.max(I_trans_vector)

z_um = z_coords_s * 1e6
r_um = r_coords_s * 1e6
r_offset_um = r_offset * 1e6
z_offset_um = z_offset * 1e6

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.plot(z_um, I_axial_scalar, color=colors['scalar'], label='Scalar', linewidth=2.5)
ax1.plot(z_um, I_axial_vector, color=colors['vector'], label='Vector (Richards-Wolf)', linewidth=2.5)
ax1.set_xlabel('z (μm)')
ax1.set_ylabel('Normalized Intensity')
ax1.set_title(f'Axial Profile (r = {r_offset_um:.2f} μm)')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(r_um, I_trans_scalar, color=colors['scalar'], label='Scalar', linewidth=2.5)
ax2.plot(r_um, I_trans_vector, color=colors['vector'], label='Vector (Richards-Wolf)', linewidth=2.5)
ax2.set_xlabel('r (μm)')
ax2.set_ylabel('Normalized Intensity')
ax2.set_title(f'Transverse Profile (z = {z_offset_um:.2f} μm)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.suptitle(f'Off-Center Profiles - NA = {NA_high} (750 nm, n = {n_medium})', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("High NA off-center: Major differences - vector effects pronounced away from axis")
```

```python
# Field Component Analysis for High NA
print("Computing field component analysis for high NA...")

z_coords_v, Ex_v, Ey_v, Ez_v = od.create_axial_profile(
    system_high, z_range=4*system_high.rayleigh_length, nz=401, r_constant=0.0, method='vector'
)
r_coords_v, Ex_v2, Ey_v2, Ez_v2 = od.create_transverse_profile(
    system_high, r_range=4*system_high.airy_radius, nr=401, z_constant=0.0, method='vector'
)

I_total_ax = np.abs(Ex_v)**2 + np.abs(Ey_v)**2 + np.abs(Ez_v)**2
I_total_tr = np.abs(Ex_v2)**2 + np.abs(Ey_v2)**2 + np.abs(Ez_v2)**2
I_Ex_ax = np.abs(Ex_v)**2 / np.max(I_total_ax)
I_Ey_ax = np.abs(Ey_v)**2 / np.max(I_total_ax)
I_Ez_ax = np.abs(Ez_v)**2 / np.max(I_total_ax)
I_Ex_tr = np.abs(Ex_v2)**2 / np.max(I_total_tr)
I_Ey_tr = np.abs(Ey_v2)**2 / np.max(I_total_tr)
I_Ez_tr = np.abs(Ez_v2)**2 / np.max(I_total_tr)

z_um = z_coords_v * 1e6
r_um = r_coords_v * 1e6

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

ax1.plot(z_um, I_Ex_ax, label='|Ex|²', linewidth=2, color='red')
ax1.plot(z_um, I_Ey_ax, label='|Ey|²', linewidth=2, color='blue')
ax1.plot(z_um, I_Ez_ax, label='|Ez|²', linewidth=2, color='green')
ax1.set_xlabel('z (μm)')
ax1.set_ylabel('Normalized |E|² Components')
ax1.set_title('Axial Field Components (r = 0)')
ax1.legend()
ax1.grid(True, alpha=0.3)

ax2.plot(r_um, I_Ex_tr, label='|Ex|²', linewidth=2, color='red')
ax2.plot(r_um, I_Ey_tr, label='|Ey|²', linewidth=2, color='blue')
ax2.plot(r_um, I_Ez_tr, label='|Ez|²', linewidth=2, color='green')
ax2.set_xlabel('r (μm)')
ax2.set_ylabel('Normalized |E|² Components')
ax2.set_title('Transverse Field Components (z = 0)')
ax2.legend()
ax2.grid(True, alpha=0.3)

plt.suptitle(f'Vector Field Components - NA = {NA_high} (750 nm, n = {n_medium})', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

focus_idx = len(z_coords_v) // 2
Ez_contribution = I_Ez_ax[focus_idx]
print(f"Ez field contribution at focus (NA = {NA_high}): {Ez_contribution*100:.1f}% of total intensity")
print("This demonstrates the importance of vector diffraction theory for high NA systems!")
```
